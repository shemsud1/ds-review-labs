## Step 1 · Types and variables

```python
# "=" stores a value in a variable. Several can be assigned at once, matched left to right.
name, age, score, active = "Abebe", 34, 0.87, True
# type() tells you what kind of value it is: str (text), int (whole number), float (decimal), bool (True/False).
print(type(name), type(age), type(score), type(active))
# Converting between types: int("42") turns text into a number; str(100) turns a number into text.
print(int("42") + 1, float("3.5") * 2, str(100) + "%")
```

## Step 2 · Strings (text)

```python
raw = "  ETH-ID-00123  "                 # text with extra spaces around it
clean = raw.strip()                       # .strip() removes spaces at both ends
print(clean.lower())                      # .lower() -> all lowercase
print(clean.replace("-", "_"))            # .replace(old, new) swaps every "-" for "_"
print(clean.split("-"))                   # .split("-") cuts at each "-" -> a list of pieces
print(clean[0:3])                         # slicing [start:end]: characters 0,1,2 (end is excluded) -> "ETH"
print(clean[-5:])                         # negative index counts from the end: last 5 characters
print(len(clean))                         # len() = number of characters
# f-string: values inside { } are filled in; :.0% formats 0.87 as a percentage with 0 decimals.
print(f"{name} is {age} years old, score {score:.0%}")
```

## Step 3 · Lists, tuples, dictionaries, sets

```python
# LIST: ordered, changeable, written with [ ].
regions = ["Oromia", "Amhara", "Sidama"]
regions.append("Afar")                    # add an item to the end
regions.sort()                            # sort alphabetically, in place
print(regions, regions[0], regions[-1], regions[1:3])   # whole list, first, last, items 1-2

# TUPLE: ordered but cannot be changed, written with ( ). Good for fixed pairs like coordinates.
coords = (9.03, 38.74)
lat, lon = coords                         # "unpacking": put each item into its own variable
print(lat, lon)

# DICTIONARY: key -> value pairs, written with { key: value }.
center = {"id": 7, "region": "Oromia", "staff": 12}
center["staff"] += 2                      # look up a value by its key and change it (12 -> 14)
center["open"] = True                     # assigning to a new key adds it
# .get(key, default) returns the default instead of crashing when the key does not exist.
print(center, center.get("phone", "n/a"))

# SET: unordered, keeps only unique values. The duplicate "walk-in" is dropped automatically.
channels = {"walk-in", "appointment", "walk-in"}
print(channels, "walk-in" in channels)    # "in" tests membership -> True
```

## Step 4 · Conditions and loops

```python
ages = [18, 42, 71, None, 30, 15]        # None means "no value"
for a in ages:                            # for-loop: run the indented block once per item
    if a is None:                         # "is None" is the correct way to test for None
        print("missing")
        continue                          # skip the rest of this round, go to the next item
    elif a < 16:                          # elif = "else if": checked only if the earlier test was False
        print(a, "too young")
        continue
    # One-line if/else (a "conditional expression"): value_if_true if condition else value_if_false
    print(a, "adult" if a >= 18 else "youth")

i = 0                                     # while-loop: repeats as long as the condition is True
while i < 3:
    print("attempt", i)
    i += 1                                # without this the loop would never end
```

## Step 5 · Functions, comprehensions, lambda

```python
def age_group(age):                      # "def" defines a reusable function with one input, age
    """Return an age band label."""       # a docstring: describes the function (shown by help())
    if age is None: return "unknown"      # "return" sends a value back and stops the function
    if age < 25:    return "16-24"
    if age < 60:    return "25-59"
    return "60+"                          # reached only if none of the tests above matched

# LIST COMPREHENSION: build a new list in one line -> [expression for item in list]
bands = [age_group(a) for a in ages]
# DICT COMPREHENSION: {key: value for item in collection}; set(bands) gives each band once
counts = {b: bands.count(b) for b in set(bands)}
# LAMBDA: a tiny one-line function. filter() keeps items where it returns True.
adults = list(filter(lambda a: a is not None and a >= 18, ages))
print(bands, counts, adults, sep="\n")   # sep="\n" prints each item on its own line

# *values collects any number of positional inputs into a tuple;
# digits=1 is a default value; **info collects extra named inputs into a dictionary.
def summarize(*values, digits=1, **info):
    return round(sum(values) / len(values), digits), info   # returns two things (as a tuple)
print(summarize(3, 5, 8, digits=2, unit="days"))
```

## Step 6 · Errors and exception handling

```python
for raw in ["12", "abc", "0"]:
    try:                                  # try: run code that might fail
        print(100 / int(raw))             # int("abc") fails; 100/0 fails
    except ValueError:                    # runs if int() could not convert the text
        print(raw, "is not a number")
    except ZeroDivisionError:             # runs if we divided by zero
        print("cannot divide by zero")
    finally:                              # always runs, whether there was an error or not
        print("checked", raw)
```

## Step 7 · Classes (object-oriented programming)
A **class** is a blueprint; an **object** is one thing built from it.

```python
class Application:                                    # define a new blueprint called Application
    def __init__(self, app_id, docs_complete, bio_quality):
        # __init__ runs automatically when you create an object. "self" = the object being built.
        self.app_id = app_id                           # store each input as an attribute of the object
        self.docs_complete = docs_complete
        self.bio_quality = bio_quality

    def is_ready(self, threshold=0.65):                # a METHOD: a function that belongs to the class
        # Ready if documents are complete AND biometric quality meets the threshold.
        return bool(self.docs_complete) and self.bio_quality >= threshold

    def __repr__(self):                                # controls how the object looks when printed
        return f"Application({self.app_id}, ready={self.is_ready()})"

class PriorityApplication(Application):                # INHERITANCE: gets everything from Application...
    def is_ready(self, threshold=0.60):                # ...but overrides this method with a lower threshold
        return super().is_ready(threshold)             # super() calls the parent class's version

# Create two objects and print them (uses __repr__). Same data, different result.
print(Application(1, True, 0.62), PriorityApplication(2, True, 0.62))
```

## Step 8 · Reading and writing files

```python
# open(file, "w") opens for writing (creates or overwrites). "with" closes the file automatically.
with open("notes.txt", "w") as f:
    f.write("center_id,staff\n7,14\n9,6\n")      # \n = new line
# open(file) with no mode = read.
with open("notes.txt") as f:
    for line in f:                                  # loop over the file one line at a time
        print(line.strip().split(","))              # remove the line break, then split at commas
```

## Step 9 · NumPy and pandas basics

```python
import numpy as np, pandas as pd          # standard nicknames

a = np.array([1, 2, 3])                  # a NumPy array: like a list, but supports fast math
b = np.array([4, 5, 6])
print(a + b)                             # element-by-element: [5 7 9] (a list would join them instead)
print(a * 2, a.mean())                   # multiply every element; average
print(a @ b)                             # @ = dot product: 1*4 + 2*5 + 3*6 = 32
print(np.linspace(0, 1, 5))              # 5 evenly spaced numbers from 0 to 1

M2 = np.array([[1, 2], [3, 4]])          # a 2-D array (matrix): list of rows
print(M2.shape, M2.T, sep="\n")          # .shape = (rows, cols); .T = transpose (swap rows/cols)

s = pd.Series([12, 7, 9], index=["A", "B", "C"])           # Series: one labelled column
d = pd.DataFrame({"center": [1, 2, 3], "staff": [12, 7, 9]})  # DataFrame: a table from a dict
print(s["B"])                            # look up a Series value by its label
display(d[d.staff > 8])                  # filter rows: d.staff > 8 makes True/False per row
```

## Step 10 · Calling a REST API and reading JSON (needs internet)

```python
import requests                          # library for talking to websites/APIs over HTTP

# GET request to GitHub's public API; timeout stops waiting after 15 seconds.
r = requests.get("https://api.github.com/repos/pandas-dev/pandas", timeout=15)
print(r.status_code)                     # 200 = OK, 404 = not found, 403 = refused/rate-limited
data = r.json()                          # convert the JSON reply into a Python dictionary
# .get() avoids a crash if a key is missing (e.g. when rate-limited).
print(data.get("full_name"), "stars:", data.get("stargazers_count"))
```
