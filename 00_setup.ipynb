## Step 1 · Check Python and core libraries

```python
# Load the libraries. If one is missing, this line fails with ModuleNotFoundError.
# sys = info about Python itself; the rest are the data science libraries.
import sys, pandas, numpy, sklearn, matplotlib, seaborn, scipy

# sys.version is a long text like "3.12.7 | packaged by Anaconda ...".
# .split() cuts it at spaces into a list; [0] takes the first item (Python counts from 0).
print("Python", sys.version.split()[0])

# Make a list of the six libraries and loop over them; "lib" holds one library each time round.
for lib in [pandas, numpy, sklearn, matplotlib, seaborn, scipy]:
    # f"..." fills in values inside { }.
    # lib.__name__ = the library's name; :12s pads it to 12 characters so the column lines up.
    # lib.__version__ = the installed version number.
    print(f"{lib.__name__:12s}", lib.__version__)
```

## Step 2 · Install the few extra libraries the labs use
Run once. Afterwards use **Kernel → Restart Kernel** so Python sees the new libraries.

```python
# %pip is a Jupyter "magic" command: it installs packages into the kernel this notebook uses.
# -q = quiet (less output). Each word after it is a package name:
#   yfinance = stock data API      dash = web dashboards     folium = maps
#   plotly = interactive charts    jupysql = the %sql magic  beautifulsoup4 + lxml = web scraping
%pip install -q yfinance dash folium plotly jupysql beautifulsoup4 lxml
```

```python
# Try to import each new library; print OK or the problem.
for name in ["yfinance", "dash", "folium", "plotly", "sql", "bs4", "lxml"]:   # "sql" = jupysql, "bs4" = beautifulsoup4
    try:
        __import__(name)                 # import a library from its name as text
        print(f"{name:10s} OK")
    except ImportError as e:             # runs only if the import failed
        print(f"{name:10s} MISSING ->", e)
```

```python
import sys, site, os, importlib            # sys = this Python; site = where libraries live; importlib = import tools

print("Kernel Python:", sys.executable)    # the exact Python this notebook runs on
user_site = site.getusersitepackages()     # your personal library folder, e.g. ~/.local/lib/python3.12/site-packages
print("Personal library folder:", user_site)

# 1. Install with THIS kernel's Python (not whatever "pip" happens to point to).
#    --user = install into your personal folder; the flag at the end hides the PATH warnings.
!{sys.executable} -m pip install -q --user --no-warn-script-location yfinance dash folium jupysql

# 2. Add the personal folder to the list of places Python searches, for this session.
if user_site not in sys.path:
    sys.path.append(user_site)
importlib.invalidate_caches()               # make Python notice the newly added folder

# 3. Make it permanent: IPython runs every file in this startup folder whenever a kernel starts.
startup = os.path.expanduser("~/.ipython/profile_default/startup")
os.makedirs(startup, exist_ok=True)         # create the folder if it doesn't exist
with open(os.path.join(startup, "00-user-site.py"), "w") as f:
    f.write("import site, sys\n"
            "p = site.getusersitepackages()\n"
            "if p not in sys.path: sys.path.append(p)\n")
print("Startup file saved.")
```
